```python
# First 5 rows
print("First 5 Rows:")
display(df.head())

# Dataset information
print("\nDataset Information:")
df.info()

# Statistics
print("\nStatistics:")
display(df.describe())

# Missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Value counts
print("\nInterns by Track:")
print(df["track"].value_counts())

print("\nSatisfaction:")
print(df["satisfied"].value_counts())
```

```python
# 1. Interns by track
sns.countplot(data=df, x="track")
plt.title("Interns by Track")
plt.show()

# 2. Quiz score distribution
sns.histplot(data=df, x="quiz_score")
plt.title("Quiz Score Distribution")
plt.show()

# 3. Tasks completed
sns.countplot(data=df, x="tasks_completed")
plt.title("Tasks Completed")
plt.show()

# 4. Hours vs quiz score
sns.scatterplot(data=df, x="hours_per_week", y="quiz_score")
plt.title("Hours per Week vs Quiz Score")
plt.show()

# 5. Days active vs satisfaction
sns.boxplot(data=df, x="satisfied", y="days_active")
plt.title("Days Active vs Satisfaction")
plt.show()
```
